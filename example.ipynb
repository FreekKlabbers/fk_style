```python
import math

import matplotlib.pyplot as plt

plt.style.use("fk_style.paper")
```

```python
# to avoid a numpy dependency, we implement a linspace function
from __future__ import division

def linspace(start, stop, n):
    """returns a generator, use list(linspace(start, stop, n)) to get a list"""
    if n == 1:
        yield stop
        return
    h = (stop - start) / (n - 1)
    for i in range(n):
        yield start + h * i
```

```python
x = list(linspace(0, 2 * math.pi, 100))
y = [math.sin(i) for i in x]

plt.plot(x, y)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.title("Sine Function")
plt.grid(True)
```
